```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
BGO_c = pd.read_table('/eos/user/l/lborriel/fcc-hardware/crystals-optical-characterization/20250224_crystal-transmittance/trasmittanza cristalli/BGO C.Sample.csv', skiprows=2,sep=',', names=['wl', 't'])



fig = plt.figure(figsize=(7, 5))

plt.plot(BGO_c.wl, BGO_c.t, label='BGO short side')
plt.xlabel('Wavelenght [nm]', size=16)
plt.ylabel('Transmittance [%]', size=16)
plt.grid()
plt.title('FCC NA - Trans. measured @ photonics lab (UNINA)', size=14)
plt.legend(fontsize=14, loc=4)
plt.xticks(np.arange(300, 750, 50))
plt.savefig('transmittanceBGOshortside.png', dpi=300, bbox_inches='tight')
plt.show()
```
